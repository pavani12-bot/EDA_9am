```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
# open source mdel no api key
# for testing purpose
import warnings
warnings.filterwarnings('ignore')
model=SentenceTransformer("all-MiniLm-L6-v2")
```

```python
sentences=[
    " I like apples.",
    "I enjoy eating apples.",
    " I drive a motorcycle."

]
```

```python
embeddings=model.encode(sentences)
```

```python
embeddings
```

```python
len(embeddings[0])
# this 384 values indicating an information i like apples
```

```python
v1=embeddings[0].reshape(-1,1) # i like apples
v2=embeddings[1].reshape(-1,1) # i enjoy eating apples
v3=embeddings[2].reshape(-1,1) # i drive a motorcycle
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(v1,v2)
```

```python
#method-1
v1=embeddings[0]
v2=embeddings[1]
def cosine_sim(v1,v2):
    return np.dot(v1,v2)/(np.linalg.norm(v1)*np.linalg.norm(v2))
print(cosine_sim(v1,v2))
```

```python
v1=embeddings[0]
v3=embeddings[2]
def cosine_sim(v1,v3):
    return np.dot(v1,v3)/(np.linalg.norm(v1)*np.linalg.norm(v3))
print(cosine_sim(v1,v3))
```

```python
#mthod-2
v1=embeddings[0]
v2=embeddings[1]
Nr=np.dot(v1,v2)
norm_v1=np.linalg.norm(v1)
norm_v2=np.linalg.norm(v2)
Dr=norm_v1*norm_v2
cosine_sim(v1,v2),Nr/Dr
```

```python
v1=embeddings[0]
v3=embeddings[2]
Nr=np.dot(v1,v3)
norm_v1=np.linalg.norm(v1)
norm_v3=np.linalg.norm(v3)
Dr=norm_v1*norm_v3
cosine_sim(v1,v3),Nr/Dr
```
